## Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv("/kaggle/input/competitions/smart-mcq-solver-challenge/train.csv")
test = pd.read_csv("/kaggle/input/competitions/smart-mcq-solver-challenge/test.csv")
```

```python
print("Train Shape:", train.shape)
print("Test Shape:", test.shape)
```

```python
import joblib

model = joblib.load(
    "/kaggle/input/datasets/venkat23f1000054/tfidf-baseline-model/tfidf_model.pkl"
)

tfidf = joblib.load(
    "/kaggle/input/datasets/venkat23f1000054/tfidf-baseline-model/tfidf_vectorizer.pkl"
)
```

```python
options = ["A","B","C","D","E"]
predictions = []
for _, row in test.iterrows():
    texts = []
    for option in options:

        texts.append(
            row["prompt"] + " [SEP] " + str(row[option])
        )
    X = tfidf.transform(texts)
    probs = model.predict_proba(X)[:,1]
    ranked = np.argsort(probs)[::-1]
    top3 = [options[i] for i in ranked[:3]]
    predictions.append(
        " ".join(top3)
    )
```

```python
sample = pd.read_csv(
    "/kaggle/input/competitions/smart-mcq-solver-challenge/sample_submission.csv"
)
sample["Prediction"] = predictions
sample.head()
```

```python
sample.to_csv("submission.csv", index=False)
```
